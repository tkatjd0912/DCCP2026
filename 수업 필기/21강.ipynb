```python
import numpy as np 
df = pd.DataFrame({
    '이름': ['강은수', '조재영', '이한솔', '뽀로로'],
    '국어': [88, np.nan, 95, 60], 
    '수학': [75, 88, np.nan, 50], '영어': [90, 79, 85, np.nan]
})
print(df)
```

```python
print(df.isna())
print(df.isna().sum())
print(df.isna().sum().sum())
print(df['국어'].notna())
```

```python
print(df.dropna())
print(df.dropna(subset=['국어']))
print(df.dropna(axis=1))
```

```python
print(df.fillna(0))
print(df.fillna({'국어':60, '수학':60, '영어':60}))

means = df[['국어', '수학', '영어']].mean()
print(df.fillna(means))
```

```python
df = pd.DataFrame({
    '이름': ['강은수', '조재영', '이한솔', '뽀로로'],
    '국어': [88, 72, 95, 60], '수학': [75, 88, 82, 50], '영어': [90, 79, 85, 70],
})

df['평균'] = (df['국어'] + df['수학'] + df['영어']) / 3
df['총점'] = df[['국어', '수학', '영어']].sum(axis=1)
df['합격'] = df['평균'] >= 70

df = df.drop(columns=['총점'])
print(df)
```
