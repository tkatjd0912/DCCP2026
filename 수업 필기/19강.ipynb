```python
import requests

params = {"postId": 1}   # 쿼리 매개변수로 필터링
r = requests.get(
    "https://jsonplaceholder.typicode.com/comments",
    params = params, timeout = 10,
)

comments: list[dict] = r.json()
print(f"댓글 {len(comments)}개")

for c in comments[:2]:
    print(f"- {c['name']} ({c['email']})")
```

```python
# 게시글의 댓글 수 세기 

import requests
from collections import Counter

BASE = "https://jsonplaceholder.typicode.com"

posts = requests.get(f"{BASE}/posts", timeout = 10).json()
comments = requests.get(f"{BASE}/comments", timeout = 10).json()

cnt = Counter(c["postId"] for c in comments)

for post in posts[:5]:
    print(f"게시글 {post['id']:>2}: 댓글 {cnt[post['id']]}개    "
          f"-- {post['title'][:30]}")
```
