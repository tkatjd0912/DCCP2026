```python
import numpy as np

x = [3, 1, 4, 1, 5, 9]
y = [2, 3, 0, 5, 1, 0]

print('mean of x    ', np.mean(x))
print('argmax of y', np.argmax(y))
```

```python
x = np.array(x)
y = np.array(y)
print('x =', x)
print('y =', y)
print(type(x))
```

```python
print('x + y =', x + y)
print('2 * x =', 2 * x)
```

```python
print('x > 2 =', x > 2)
print('x ** 2 =', x ** 2)
```

```python
import numpy as np 
import timeit

N = 1_000_000
xs_list = list(range(N))
xs_np   = np.arange(N)

t1 = timeit.timeit("[x*x for x in xs_list]", globals=globals(), number = 1)
t2 = timeit.timeit("xs_np ** 2",             globals=globals(), number=1)

print(f"리스트 컴프리헨션: {t1*1000:7.1f} ms")
print(f"numpy 벡터화    : {t2*1000:7.1f} ms")
print(f"속도 비         : {t1/t2:.1f}배")
```

```python
# 불 인덱싱: 조건식

scores = np.array([88, 72, 95, 60, 81, 77, 90])

print(scores >= 80)         # 불 배열
print(scores[scores>=80])   # 필터링 (80점 이상만)

print(scores[(scores >= 70) & (scores < 90)])
```

```python
# 팬시 인덱싱 

names = np.array(['강은수', '조재영', '이한솔', '뽀로로'])

print(names[[0, 3]])
print(names[[3, 0, 3, 0]])
```

```python
# 팬시 인덱싱 활용

names = np.array(['강은수', '조재영', '이한솔', '뽀로로'])
scores = np.array([88, 72, 95, 60])

order = scores.argsort()
print(order)
print(names[order])
print(names[order][::-1])
```

```python
# 다차원 팬시 인덱싱

a = np.arange(10)
a[[1, 3, 5]] = 0
print(a)            # 팬시 인덱싱은 복사본을 돌려줌

scores = np.array([88, 45, 95, 30, 77])
scores[scores < 60] = 60
print(scores)
```
