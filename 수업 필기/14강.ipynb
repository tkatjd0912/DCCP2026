▶ 절대 경로: 루트부터 시작하는 전체 주소
▶ 상대 경로: 현재 작업 디렉터리를 기준으로 한 주소. 실습 코드에 유리함

▶ open()
- 사용법: f = open(file, mode, encoding)
- 모드: 'r'(읽기), 'w'(덮어쓰기), 'a'(추가하기), 'x'(새 파일 생성), 'b'(다른 모드와 결합), 't'(텍스트 모드), '+'(읽기쓰기)
* 'w'로 열면 기존 파일 내용이 즉시 사라짐
* close(): 파일을 다 쓰고 나면 반드시 close()로 반납해야 함. 데이터가 디스크에 기록되지 않을 수 있다
* with 문법: 파일을 자동으로 닫아줌
- 문법: with open(file, mode, encoding) as f:

▶ 읽기 메소드 (‘r’)
- f.read(): 전체를 한번에
- f.realine(): 한 줄씩
- f.readlines(): 줄들의 리스트 

▶ 쓰기 메소드 (‘w’)
- f.write(s): 문자열 s를 파일에 추가. 자동 줄바꿈이 적용되지 않으므로 \n을 써주어야 함
- f.writelines(list): 문자열 list를 파일에 추가. 자동 줄바꿈이 적용되지 않으므로 \n을 써주어야 함



```python
# 텍스트 파일 생성
with open ('poem.txt', 'w', encoding='utf-8') as f:
    f.write("죽는 날까지 하늘을 우러러\n")
    f.write("한 점 부끄럼이 없기를\n")
    f.write("잎새에 이는 바람에도\n")
    f.write("나는 괴로워했다\n")
```

```python
# 한 줄씩 처리 
f = open("poem.txt", "r", encoding='utf-8')
for i, line in enumerate(f, start=1):
    print(f"{i}행: {line.rstrip()}")
f.close()
```

```python
# 단어 빈도 계산
text = "the cat sat on the mat the cat slept"
words = text.split()
freq = {}
for w in words:
    freq[w] = freq.get(w, 0) + 1

with open("freq.txt", "w", encoding="utf-8") as f:
    for word, count in sorted(freq.items()):
        f.write(f"{word}\t{count}\n")
```

```python
# 행을 리스트로 쓰기
import csv

rows = [
    ["언어", "사용자수(백만)", "어족"],
    ["한국어", 81, "한국어족"],
    ["영어", 1500, "인도유럽어족"],
    ["스페인어", 600, "인도유럽어족"]
]

with open("languages.csv", "w", encoding="utf-8", newline="") as f:
    writer = csv.writer(f)      # newline: 빈 줄이 끼어드는 것을 방지
    writer.writerows(rows)
```

```python
import json 

corpus = {
    "이름": "한국어 일상 대화 코퍼스", 
    "발화": [
        {"화자": "A", "내용": "안녕?", "시간": 0.5}, 
        {"화자": "B", "내용": "오랜만!", "시간": 1.2}, 
        {"화자": "C", "내용": "잘 지냈어?", "시간": 2.4},
    ]
}

with open("corpus.json", "w", encoding = "utf-8") as f:
    json.dump(corpus, f, ensure_ascii = False, indent = 2)
    
with open("corpus.json", "r", encoding = "utf-8") as f:
    data = json.load(f)
    
for utt in data["발화"]:
    print(f"{utt["시간"]:.1f}s [{utt["화자"]}] {utt["내용"]}")
```

```python
# 실습 1 - speech.txt 파일이 없기 때문에 출력 결과는 오류 

freq = {}
with open ("speech.txt", "r", encoding = "utf-8") as f:
    for line in f:
        for word in line.lower().split():
            word = word.strip(".,!?\"'")
            if word:
                freq[word] = freq.get(word, 0) + 1
                
ranked = sorted(freq.items(), key = lambda x: x[1], reverse = True)
for word, count in ranked[:5]:
    print(f"{word}: {count}")
```

```python
# 실습 2

import csv, json

students = []
with open("students.csv", "r", encoding = "utf-8") as f:
    reader = csv.DictReader(f)
    for row in reader:
        students.append(row)

with open("students.json", "w", encoding = "utf-8") as f:
    json.dump(students, f, ensure_ascii= False, indent = 2)
```

```python
# 실습 3

import json 

stats = []

with open("poem.txt", "r", encoding = "utf-8") as f:
    for i, line in enumerate(f, start = 1):
        line = line.strip()
        stats.append({
            "행번호": i, "내용": line, 
            "글자수": len(line), "어절수": len(line.split()),
        })
    
with open("poem.txt", "w", encoding = "utf-8") as f:
    json.dump(stats, f, ensure_ascii = False, indent = 2)
```
