```python
# 예제 1: 다중 공백 정규화
import re
def normalize_whitespace(text: str) -> str:
    text = re.sub(r"[ \t]+", " ", text)
    text = re.sub(r"\n{3,}", "\n\n", text)
    return text.strip()

normalize_whitespace("안녕  세상\n\n\n파이썬\t\t재미")
```

```python
# 예제 2: HTML 태그 제거 
html = "<p>안녕 <b>세상</b>!</p><p>두 번째 <i>문단</i></p>"

re.sub(r"<[^>]+>", "", html)
```

```python
# 예제 3: 개인정보 마스킹
def mask_pii(text: str) -> str:
    text = re.sub(r"\d{2,4}-\d{3,4}-\d{4}", "[전화]", text)
    text = re.sub(r"[\w.+-]+@[\w-]+\.[\w.-]+", "[이메일]", text)
    text = re.sub(r"\d{6}-[1-4]\d{6}", "[주민등록]", text)
    return text

essay = """문의: mam3b@snu.ac.kr (010-1234-5678)
주민등록번호 950101-1234567 작성 금지."""
print(mask_pii(essay))
```

```python
# 예제 4: 한국어 텍스트 정제 파이프라인
def clean_korean(text: str) -> str:
    text = re.sub(r"https?://\S+"," ", text)
    text = re.sub(r"@\w+", " ", text)
    text = re.sub(r"#\w+", " ", text)
    text = re.sub(r"[^가-힣\s]+", " " , text)
    text = re.sub(r"\s+", " ", text).strip()
    return text
      
tweet = "오늘 #봄날씨 진짜 최고! @friend 보러 가자 https://x.com/abc"
print(clean_korean(tweet))
```

```python
# 예제 5: 마크다운에서 링크 추출
md = """공식 [파이썬](https://python.org)과
강의 [eTL](https://etl.snu.ac.kr/)을 참고하라."""

pattern = re.compile(r"\[(?P<text>[^\]]+)\]\((?P<url>[^)]+)\)")

for m in pattern.finditer(md):
    print(f"{m['text']:8} --> {m['url']}")
```

```python
# 예제 6: 동의어 일괄 통일
synonyms = {
    "컴퓨터": "PC",
    "휴대전화": "핸드폰",
    "전자우편": "이메일",
}

pattern = re.compile("|".join(re.escape(k) for k in synonyms))

text = "컴퓨터로 전자우편을 쓰고 휴대전화로 확인한다."
result = pattern.sub(lambda m: synonyms[m.group()], text)
print(result)
```

```python
# 예제 7: 인용 부호 통일
text = """은수는 "안녕"이라고, 재영이는 '잘 가'라고 했다."""

text = re.sub("\u201c([^\u201d]+)\u201d", r'"\1"', text)
text = re.sub("\u2018([^\u2019]+)\u2019", r"'\1'", text)
print(text)
```

```python
# 예제 8: 한국어 어미 분석
text = "걸으면서, 먹으면서, 자면서, 가면서 살았다."

pattern = re.compile(r"(\w+?)((?:으)?면서)")

for m in pattern.finditer(text):
    print(f"어간 후보: {m.group(1):4}  어미: {m.group(2)}")
```
