```python
import os
from dotenv import load_dotenv
load_dotenv()
KEY: str = os.environ["KOREAN_DICT_KEY"]
print(KEY[:4] + "***")
```

Q1

```python
# Q1 - (a)

import requests

def search_word(q: str, num: int = 10, start: int = 1) -> dict:
    url = "https://opendict.korean.go.kr/api/search"
    params = {
        "key": KEY,
        "q": q,
        "req_type": "json",
        "num": num,
        "start": start,
        "type1": "word",
    }
    r = requests.get(url, params=params, timeout=10)
    r.raise_for_status()
    return r.json()
```

```python
# Q1 - (b)

import json

data = search_word("김치")
print(json.dumps(data, ensure_ascii=False, indent=2)[:400])

""""설명: ensure_ascii=False를 빼면 한글이 \uc6b0\ub9ac\ub9d0\uc0d8처럼 유니코드 이스케이프 시퀀스로 출력되어 사람이 읽기 어렵다."""
```

```python
# Q1 - (c)

# (i)
total: int = data["channel"]["total"]
n: int = data["channel"]["num"]
print(f"총 {total}건, 이 페이지 {n}건")

# (ii)
items: list[dict] = data["channel"]["item"]
for item in items[:5]:
    word: str = item.get("word", "")
    pos: str = item.get("pos", "품사 없음")
    sense: str = item["sense"][0]["definition"][:40]
    print(f"{word} ({pos}) --> {sense}")
```
